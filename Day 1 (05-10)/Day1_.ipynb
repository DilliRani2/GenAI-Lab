```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Job_Title"
)

plt.title("Distribution of AI/ML Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Number of Jobs")

plt.xticks(rotation=30)

plt.show()

```

```python
plt.figure(figsize=(6, 5))

sns.countplot(
    data=df,
    x="Python"
)

plt.title("Python Requirement in AI/ML Jobs")

plt.show()
```

```python
skills = ["Python", "NumPy", "Pandas", "Scikit_Learn"]

skill_counts = {}

for skill in skills:
    skill_counts[skill] = (df[skill] == "Yes").sum()

```

```python
skill_df = pd.DataFrame(
    list(skill_counts.items()),
    columns=["Skill", "Job_Count"]
)

skill_df
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=skill_df,
    x="Skill",
    y="Job_Count"
)

plt.title("AI/ML Skills Required Across Job Records")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Salary_LPA",
    bins=10,
    kde=True
)

plt.title("Salary Distribution")
plt.xlabel("Salary (LPA)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Experience_Years",
    y="Salary_LPA",
    hue="Job_Title",
    s=100
)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (LPA)")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Job_Title",
    y="Salary_LPA"
)

plt.title("Salary Distribution by Job Role")

plt.xticks(rotation=30)

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Education",
    y="Salary_LPA"
)

plt.title("Education vs Salary")

plt.show()
```
